```python
# Scrapping
jobs = []

for role in job_roles:

    role_url = role.lower().replace(" ", "-")
    role_query = role.lower().replace(" ", "+")

    for page in range(1, 51):

        if page == 1:
            url = f"https://www.naukri.com/{role_url}-jobs?k={role_query}"
        else:
            url = f"https://www.naukri.com/{role_url}-jobs-{page}?k={role_query}"

        driver.get(url)
        time.sleep(2)

        soup = BeautifulSoup(driver.page_source, "html.parser")

        job_cards = soup.find_all("div", class_="cust-job-tuple")

        for job in job_cards:

            title_tag = job.find("a", class_="title") 
            if not title_tag: 
                continue
            company_tag = job.find("a", class_="comp-name")
            rating_tag = job.find("span", class_="main-2")
            review_tag = job.find("a", class_="review")
            experience_tag = job.find("span", class_="expwdth")
            salary_tag = job.find("span", class_="sal-wrap")
            location_tag = job.find("span", class_="locWdth")
            description_tag = job.find("span", class_="job-desc")
            posted_tag = job.find("span", class_="job-post-day")

            skills = job.find_all("li", class_="tag-li")
            if not title_tag.get("href"):
                print(f"Missing href | Role: {role} | Page: {page}")
                print(title_tag)
                continue
            jobs.append({
                "Role Searched": role,
                "Job Title": title_tag.text.strip() if title_tag else None,
                "Company": company_tag.text.strip() if company_tag else None,
                "Rating": rating_tag.text.strip() if rating_tag else None,
                "Reviews": review_tag.text.strip() if review_tag else None,
                "Experience": experience_tag.text.strip() if experience_tag else None,
                "Salary": salary_tag.text.strip() if salary_tag else "Not Disclosed",
                "Location": location_tag.text.strip() if location_tag else None,
                "Description": description_tag.text.strip() if description_tag else None,
                "Skills": ", ".join(skill.text.strip() for skill in skills),
                "Posted": posted_tag.text.strip() if posted_tag else None,
                "Job Link": title_tag.get("href") if title_tag else None
            })

        print(f"{role} | Page {page} | Total Jobs: {len(jobs)}")
```

```python
df = pd.DataFrame(jobs)

print(df.shape)

df.to_csv("naukri_jobs.csv", index=False)
```

### **Explanation -**
In the first notebook, I developed an automated web scraping pipeline to collect real-world job market data from Naukri.com. Since the objective of the project was to analyse the Indian job market for data-related roles, I selected six relevant search categories: Data Analyst, Data Scientist, Data Engineer, Machine Learning Engineer, AI Engineer, and Business Analyst. These roles cover a broad spectrum of the analytics and AI industry, making the dataset representative of current hiring trends.

Initially, I explored using the Requests library with BeautifulSoup, but inspection of the returned HTML revealed that Naukri loads job listings dynamically using JavaScript (React/Next.js). As a result, the required job information was not present in the raw HTML, making traditional web scraping ineffective. To overcome this, I used Selenium WebDriver, which automates a real browser, allowing the page to load completely before extracting its HTML. Once the page was rendered, BeautifulSoup was used to efficiently parse the HTML because it is faster and simpler than locating every element directly through Selenium.

The scraper automatically navigated through multiple pages for each job role and extracted important attributes such as job title, company name, company rating, number of reviews, experience required, salary, location, job description, required skills, posting time, and job link. Some companies did not display ratings or salaries, so the scraper intentionally preserved these missing values instead of replacing them, ensuring that the dataset accurately reflected the information available on the website. The final output consisted of 4,136 job postings stored as a CSV file, providing a comprehensive real-world dataset for further preprocessing, analysis, and predictive modelling.
